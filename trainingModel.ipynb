```python
from scipy.io import arff
import pandas as pd

data, meta = arff.loadarff("1year.arff")

df = pd.DataFrame(data)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
for col in df.columns:
    if df[col].dtype == object:
        df[col] = df[col].apply(
            lambda x: x.decode("utf-8")
            if isinstance(x, bytes)
            else x
        )

df = df.apply(pd.to_numeric, errors="coerce")

print(df.head())
print(df.dtypes)
```

```python
print("Rows and columns:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nLast column:")
print(df.columns[-1])

print("\nTarget distribution:")
print(df.iloc[:, -1].value_counts())
```

```python
print("Rows and columns:", df.shape)
print(df.columns.tolist())
print(df.iloc[:, -1].value_counts())
```

```python
print("Total missing values:", df.isnull().sum().sum())

print("\nMissing values by column:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
X = df.drop("class", axis=1)

y = df["class"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts())

print("\nTesting target:")
print(y_test.value_counts())
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)

X_test_imputed = imputer.transform(X_test)

print("Missing values handled successfully!")
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_imputed,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(
    X_train_smote,
    y_train_smote
)

print("✅ Random Forest model trained successfully!")
```

```python
y_pred = rf_model.predict(X_test_imputed)

y_probability = rf_model.predict_proba(X_test_imputed)[:, 1]

print("Predictions completed!")
print("First 10 predictions:")
print(y_pred[:10])

print("\nFirst 10 bankruptcy probabilities:")
print(y_probability[:10])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    zero_division=0
)

roc_auc = roc_auc_score(
    y_test,
    y_probability
)

print("================================")
print("   MODEL PERFORMANCE")
print("================================")

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title("Corporate Bankruptcy Prediction - Confusion Matrix")

plt.show()
```

```python
feature_importance = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
)

feature_importance = feature_importance.sort_values(
    ascending=False
)

print("Top 15 important financial attributes:")
print(feature_importance.head(15))
```

```python
plt.figure(figsize=(10, 7))

feature_importance.head(15).sort_values().plot(
    kind="barh"
)

plt.xlabel("Importance")
plt.ylabel("Financial Attribute")

plt.title(
    "Top Financial Factors Influencing Bankruptcy"
)

plt.tight_layout()

plt.show()
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = [0.50, 0.45, 0.40, 0.35, 0.30, 0.25, 0.20]

print("THRESHOLD COMPARISON")
print("=" * 70)

for threshold in thresholds:

    predictions = (
        y_probability >= threshold
    ).astype(int)

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    print(
        f"Threshold: {threshold:.2f} | "
        f"Precision: {precision:.4f} | "
        f"Recall: {recall:.4f} | "
        f"F1: {f1:.4f}"
    )
```

```python
import numpy as np
from sklearn.metrics import f1_score

threshold_values = np.arange(
    0.10,
    0.71,
    0.01
)

best_threshold = 0
best_f1 = 0

for threshold in threshold_values:

    predictions = (
        y_probability >= threshold
    ).astype(int)

    current_f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    if current_f1 > best_f1:

        best_f1 = current_f1
        best_threshold = threshold


print("Best threshold:", round(best_threshold, 2))
print("Best F1 Score:", round(best_f1, 4))
```

```python
optimal_predictions = (
    y_probability >= best_threshold
).astype(int)

optimal_precision = precision_score(
    y_test,
    optimal_predictions,
    zero_division=0
)

optimal_recall = recall_score(
    y_test,
    optimal_predictions,
    zero_division=0
)

optimal_f1 = f1_score(
    y_test,
    optimal_predictions,
    zero_division=0
)

print("======================================")
print("OPTIMIZED BANKRUPTCY CLASSIFICATION")
print("======================================")

print(
    "Threshold :",
    round(best_threshold, 2)
)

print(
    "Precision :",
    round(optimal_precision, 4)
)

print(
    "Recall    :",
    round(optimal_recall, 4)
)

print(
    "F1 Score  :",
    round(optimal_f1, 4)
)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

cm = confusion_matrix(
    y_test,
    optimal_predictions
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues"
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title(
    "Optimized Bankruptcy Prediction - Confusion Matrix"
)

plt.show()
```
